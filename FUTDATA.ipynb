```python
loss_events = ["Miscontrol", "Dispossessed", "Dribble Lost"]
shots = df[df["type.name"] == "Shot"].copy()
df_losses = df[df["type.name"].isin(loss_events)].copy()
df_losses = df_losses.dropna(subset=["location"])
```

```python
def time_to_next_shot(loss_row, shots_df):
    same_game_shots = shots_df[
        (shots_df["minute"] > loss_row["minute"]) |
        (
            (shots_df["minute"] == loss_row["minute"]) &
            (shots_df["second"] > loss_row["second"])
        )
    ]

    if same_game_shots.empty:
        return None

    next_shot = same_game_shots.iloc[0]

    return (next_shot["minute"] * 60 + next_shot["second"]) - \
           (loss_row["minute"] * 60 + loss_row["second"])
```

```python
df_losses["time_to_shot"] = df_losses.apply(
    lambda row: time_to_next_shot(row, shots),
    axis=1
)
```

```python
df_losses["danger"] = 1 / (df_losses["time_to_shot"] + 1)
```

```python
# Extraemos la x y la y de la columna 'location' de StatsBomb
df_losses['x'] = df_losses['location'].apply(lambda loc: loc[0] if isinstance(loc, list) else None)
df_losses['y'] = df_losses['location'].apply(lambda loc: loc[1] if isinstance(loc, list) else None)

# Eliminamos filas que no tengan coordenadas por si acaso
df_losses = df_losses.dropna(subset=['x', 'y'])
```

```python
import os
import json
import glob
import pandas as pd

# 1. Miramos qué archivos de eventos tienes en tu ordenador
archivos_eventos = glob.glob("open-data/data/events/*.json")
if not archivos_eventos:
    archivos_eventos = glob.glob("data/events/*.json")

match_ids_locales = [os.path.basename(f).replace('.json', '') for f in archivos_eventos]

if not match_ids_locales:
    print("❌ Tu carpeta 'events' está vacía. No hay datos de ningún partido aún.")
else:
    print(f"Tienes {len(match_ids_locales)} partidos en tu carpeta de eventos. Buscando sus nombres...\n")
    
    # 2. Buscamos esos IDs en la carpeta de partidos
    rutas_matches = glob.glob("open-data/data/matches/**/*.json", recursive=True) + glob.glob("data/matches/**/*.json", recursive=True)
    
    partidos_encontrados = []
    
    for ruta in rutas_matches:
        with open(ruta, "r", encoding="utf-8") as f:
            try:
                partidos = json.load(f)
                for p in partidos:
                    mid = str(p['match_id'])
                    if mid in match_ids_locales:
                        partidos_encontrados.append({
                            "ID Partido (Match ID)": mid,
                            "Competición": p["competition"]["competition_name"],
                            "Temporada": p["season"]["season_name"],
                            "Partido": f"{p['home_team']['home_team_name']} vs {p['away_team']['away_team_name']}"
                        })
            except:
                continue
                
    if partidos_encontrados:
        df_mis_partidos = pd.DataFrame(partidos_encontrados).drop_duplicates()
        print("🏆 ESTOS SON LOS PARTIDOS REALES QUE TIENES EN TU CARPETA 'EVENTS': 🏆")
        display(df_mis_partidos)
    else:
        print("Tienes los archivos de eventos, pero no encuentro sus nombres en la carpeta 'matches'.")
```

```python
import pandas as pd

# 1. Aseguramos el orden cronológico estricto del dataset
df_ord = df_barsa_events.sort_values(by=["match_id", "period", "minute", "second"]).reset_index(drop=True)

# 2. Identificamos los índices de todos los tiros del dataset
indices_tiros = df_ord[df_ord["type.name"] == "Shot"].index.tolist()

def obtener_lado(loc):
    if not isinstance(loc, list) or len(loc) < 2: return "Desconocido"
    y = loc[1]
    return "Lado Izquierdo" if y < 26.66 else ("Lado Derecho" if y > 53.33 else "Zona Central")

lados_criticos = []

# 3. Rastreamos hacia atrás desde cada tiro
for idx in indices_tiros:
    tiro = df_ord.iloc[idx]
    eq_tiro = tiro["team.name"]
    periodo = tiro["period"]
    m_id = tiro["match_id"]
    t_tiro = (tiro.get("minute", 0) * 60) + tiro.get("second", 0)
    
    for i in range(1, 35):
        idx_prev = idx - i
        if idx_prev < 0: break
            
        ev = df_ord.iloc[idx_prev]
        if ev["match_id"] != m_id or ev["period"] != periodo: break
            
        t_prev = (ev.get("minute", 0) * 60) + ev.get("second", 0)
        
        # Filtro estricto: menos de 30 segundos
        if (t_tiro - t_prev) > 30: break
            
        # Comprobamos si es pérdida del Barça y el tiro es del rival
        es_perdida = (ev["type.name"] in ["Dispossessed", "Miscontrol"]) or (
            (ev["type.name"] == "Pass") & (pd.notna(ev.get("pass.outcome.name")))
        )
        
        if es_perdida and "Barcelona" in ev["team.name"] and eq_tiro != ev["team.name"]:
            lados_criticos.append(obtener_lado(ev.get("location")))
            break

# 4. Mostramos el resultado
df_critico = pd.DataFrame(lados_criticos, columns=["Lado"])
df_res = pd.DataFrame({
    "Pérdidas Críticas (<30s)": df_critico["Lado"].value_counts(),
    "Porcentaje del Peligro (%)": (df_critico["Lado"].value_counts(normalize=True) * 100).round(2)
})
display(df_res)
```

```python
import sys
!{sys.executable} -m pip install scikit-learn xgboost
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, roc_auc_score
from xgboost import XGBClassifier

# 1. Asegurar orden cronológico
df_ord = df_barsa_events.sort_values(by=["match_id", "period", "minute", "second"]).reset_index(drop=True)

# 2. Identificar los índices de todos los goles rivales
indices_goles_rival = df_ord[
    (df_ord["type.name"] == "Shot") & 
    (df_ord["shot.outcome.name"] == "Goal") & 
    (~df_ord["team.name"].str.contains("Barcelona", na=False))
].index.tolist()

# 3. Mapear marcas de tiempo de los goles rivales por partido y periodo
goles_dict = {}
for idx in indices_goles_rival:
    row = df_ord.iloc[idx]
    m_id, per = row["match_id"], row["period"]
    t_gol = (row.get("minute", 0) * 60) + row.get("second", 0)
    goles_dict.setdefault((m_id, per), []).append(t_gol)

# 4. Filtrar todas las recepciones (toques iniciales) de Messi y Neymar
df_rec = df_ord[
    (df_ord["type.name"] == "Ball Receipt*") & 
    (df_ord["player.name"].str.contains("Messi|Neymar", case=False, na=False)) &
    (df_ord["location"].notna())
].copy()

# 5. Ingeniería de Variables (Features) y Variable Objetivo (Target)
features = []
targets = []

for idx, row in df_rec.iterrows():
    m_id, per = row["match_id"], row["period"]
    t_rec = (row.get("minute", 0) * 60) + row.get("second", 0)
    
    # ¿Hubo un gol rival en los siguientes 20 segundos en este mismo partido/tiempo?
    es_gol_en_contra = 0
    if (m_id, per) in goles_dict:
        for t_gol in goles_dict[(m_id, per)]:
            if 0 < (t_gol - t_rec) <= 20: # Ventana crítica de 20 segundos
                es_gol_en_contra = 1
                break
                
    # Extraemos variables clave del momento de la recepción
    loc = row["location"]
    under_pressure = 1 if row.get("under_pressure") == True else 0
    es_messi = 1 if "Messi" in str(row["player.name"]) else 0
    
    features.append([
        loc[0],          # Coordenada X (Largo del campo)
        loc[1],          # Coordenada Y (Ancho del campo)
        t_rec,           # Segundo del partido (fatiga/minuto)
        under_pressure,  # ¿Recibe bajo presión?
        es_messi         # 1 si es Messi, 0 si es Neymar
    ])
    targets.append(es_gol_en_contra)

# 6. Crear matrices para el modelo
X = pd.DataFrame(features, columns=["loc_x", "loc_y", "timestamp", "under_pressure", "es_messi"])
y = np.array(targets)

print(f"📋 Datos procesados: {len(y)} recepciones analizadas.")
print(f"🚨 Recepciones críticas que terminaron en gol en contra: {sum(y)}")

# 7. Dividir en Entrenamiento y Test
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

# Calcular balance de pesos debido al desequilibrio extremo de goles
ratio_neg_pos = (len(y_train) - sum(y_train)) / sum(y_train) if sum(y_train) > 0 else 1

# 8. Entrenar el modelo XGBoost
model = XGBClassifier(
    n_estimators=100,
    max_depth=4,
    learning_rate=0.05,
    scale_pos_weight=ratio_neg_pos, # Ajuste vital para datos desequilibrados
    eval_metric="logloss",
    random_state=42
)
model.fit(X_train, y_train)

# 9. Evaluación del Modelo
y_pred = model.predict(X_test)
print("\n📊 INFORME DE RENDIMIENTO DEL MODELO:")
print("-" * 55)
print(classification_report(y_test, y_pred, target_names=["Seguro", "Peligro Gol"]))
```

```python
import pandas as pd

# 1. Aseguramos el orden cronológico
df_ord = df_barsa_events.sort_values(by=["match_id", "period", "minute", "second"]).reset_index(drop=True)

# 2. Filtramos recepciones de Neymar en todo el largo de su carril izquierdo (y <= 26.66)
df_ney_rec = df_ord[
    (df_ord["type.name"] == "Ball Receipt*") &
    (df_ord["player.name"].str.contains("Neymar", case=False, na=False)) &
    (df_ord["location"].notna())
].copy()

df_ney_rec["x"] = df_ney_rec["location"].apply(lambda loc: loc[0])
df_ney_rec["y"] = df_ney_rec["location"].apply(lambda loc: loc[1])

# Filtro únicamente para su carril izquierdo, sin límite en el largo X
df_ney_carril = df_ney_rec[df_ney_rec["y"] <= 26.66].copy()

# Nueva segmentación en 4 zonas a lo largo de todo el campo (0 a 120 metros)
def clasificar_largo_total(x):
    if x <= 30: return "1. Propio Campo Bajo (0-30m)"
    elif x <= 60: return "2. Propio Campo Alto (30-60m)"
    elif x <= 90: return "3. Campo Rival - Creación (60-90m)"
    else: return "4. Campo Rival - Finalización (90-120m)"

df_ney_carril["zona_largo"] = df_ney_carril["x"].apply(clasificar_largo_total)

# 3. Rastrear la acción INMEDIATA posterior a la recepción
desenlaces_totales = []

for idx, row in df_ney_carril.iterrows():
    m_id = row["match_id"]
    per = row["period"]
    
    # Buscamos los siguientes eventos para ver qué hace Neymar con el balón
    futuro = df_ord.iloc[idx+1 : idx+5]
    futuro_mismo_tiempo = futuro[(futuro["match_id"] == m_id) & (futuro["period"] == per)]
    
    accion_inmediata = "Desconducción/Pérdida indeterminada"
    
    for _, f_ev in futuro_mismo_tiempo.iterrows():
        if f_ev.get("player.name") != row["player.name"]:
            if "Barcelona" not in str(f_ev.get("team.name")):
                accion_inmediata = "Pérdida (Intercepción Rival)"
            break
            
        tipo = f_ev["type.name"]
        
        if tipo == "Pass":
            if pd.isna(f_ev.get("pass.outcome.name")):
                accion_inmediata = "Pase Completado"
            else:
                accion_inmediata = "Pase Fallado (Pérdida)"
            break
        elif tipo == "Dribble":
            if f_ev.get("dribble.outcome.name") == "Complete":
                accion_inmediata = "Regate Completado"
            else:
                accion_inmediata = "Regate Fallado (Pérdida)"
            break
        elif tipo in ["Miscontrol", "Dispossessed"]:
            accion_inmediata = "Pérdida (Robo/Mal Control)"
            break
        elif tipo == "Shot":
            accion_inmediata = "Tiro a Puerta"
            break

    desenlaces_totales.append({
        "Zona Horizontal (X)": row["zona_largo"],
        "Desenlace": accion_inmediata
    })

df_desenlaces_total = pd.DataFrame(desenlaces_totales)

# 4. Mostramos la matriz cruzada de todo el largo del carril
print("📊 COMPORTAMIENTO DE NEYMAR A LO LARGO DE TODO SU CARRIL IZQUIERDO (2014-2017)")
print("-" * 85)
matriz_completa = pd.crosstab(
    df_desenlaces_total["Zona Horizontal (X)"], 
    df_desenlaces_total["Desenlace"], 
    normalize="index"
) * 100

display(matriz_completa.round(2).astype(str) + " %")
```

## 2. MINUTO Y RESULTAO

```python
import pandas as pd

# 1. Ordenamos cronológicamente todo el dataset
df_ord = df_barsa_events.sort_values(by=["match_id", "period", "minute", "second"]).reset_index(drop=True)

# 2. Identificamos todos los goles del partido para construir el marcador en vivo
df_ord["es_gol_barca"] = (df_ord["type.name"] == "Shot") & \
                         (df_ord["shot.outcome.name"] == "Goal") & \
                         (df_ord["team.name"].str.contains("Barcelona", na=False))

df_ord["es_gol_rival"] = (df_ord["type.name"] == "Shot") & \
                         (df_ord["shot.outcome.name"] == "Goal") & \
                         (~df_ord["team.name"].str.contains("Barcelona", na=False))

# Creamos el contador acumulado de goles por partido
df_ord["goles_barca_acum"] = df_ord.groupby("match_id")["es_gol_barca"].cumsum()
df_ord["goles_rival_acum"] = df_ord.groupby("match_id")["es_gol_rival"].cumsum()

# ⚠️ Ajuste crítico: El gol modifica el marcador DESPUÉS del evento, no antes. 
# Para saber el marcador exacto CUANDO Neymar recibe, desplazamos el acumulado una fila.
df_ord["marcador_barca"] = df_ord.groupby("match_id")["goles_barca_acum"].shift(1).fillna(0).astype(int)
df_ord["marcador_rival"] = df_ord.groupby("match_id")["goles_rival_acum"].shift(1).fillna(0).astype(int)
df_ord["diferencia_marcador"] = df_ord["marcador_barca"] - df_ord["marcador_rival"]

# 3. Filtramos el carril izquierdo de Neymar (y <= 26.66)
df_ney_rec = df_ord[
    (df_ord["type.name"] == "Ball Receipt*") &
    (df_ord["player.name"].str.contains("Neymar", case=False, na=False)) &
    (df_ord["location"].notna())
].copy()

df_ney_rec["x"] = df_ney_rec["location"].apply(lambda loc: loc[0])
df_ney_rec["y"] = df_ney_rec["location"].apply(lambda loc: loc[1])
df_ney_carril = df_ney_rec[df_ney_rec["y"] <= 26.66].copy()

# 4. Clasificamos el estado del partido en el momento de la recepción
def clasificar_estado(dif):
    if dif > 0: return "Ganando"
    elif dif < 0: return "Perdiendo"
    else: return "Empatando"

df_ney_carril["estado_partido"] = df_ney_carril["diferencia_marcador"].apply(clasificar_estado)

# Clasificamos el tramo del partido (Minuto)
def clasificar_tramo(minuto):
    if minuto <= 15: return "1. Inicio (0'-15')"
    elif minuto <= 45: return "2. Primera Parte (16'-45')"
    elif minuto <= 75: return "3. Segunda Parte (46'-75')"
    else: return "4. Tramo Final (76'-90+')"

df_ney_carril["tramo_minuto"] = df_ney_carril["minute"].apply(clasificar_tramo)

# 5. Rastreamos qué hace inmediatamente después (mismo código de desenlaces corto)
desenlaces_contexto = []
for idx, row in df_ney_carril.iterrows():
    m_id, per = row["match_id"], row["period"]
    futuro = df_ord.iloc[idx+1 : idx+5]
    futuro_mismo_tiempo = futuro[(futuro["match_id"] == m_id) & (futuro["period"] == per)]
    
    accion = "Pérdida/Otros"
    for _, f_ev in futuro_mismo_tiempo.iterrows():
        if f_ev.get("player.name") != row["player.name"]:
            if "Barcelona" not in str(f_ev.get("team.name")): accion = "Pérdida (Robo Rival)"
            break
        tipo = f_ev["type.name"]
        if tipo == "Pass":
            accion = "Pase Completado" if pd.isna(f_ev.get("pass.outcome.name")) else "Pase Fallado"
            break
        elif tipo == "Dribble":
            accion = "Regate Completado" if f_ev.get("dribble.outcome.name") == "Complete" else "Regate Fallado"
            break
        elif tipo in ["Miscontrol", "Dispossessed"]:
            accion = "Pérdida (Robo Rival)"
            break

    desenlaces_contexto.append(accion)

df_ney_carril["desenlace"] = desenlaces_contexto

# 6. Analizamos el impacto del Marcador en su juego
print("📊 ¿CÓMO CAMBIA EL JUEGO DE NEYMAR SEGÚN EL MARCADOR EN VIVO?")
print("-" * 75)
matriz_marcador = pd.crosstab(df_ney_carril["estado_partido"], df_ney_carril["desenlace"], normalize="index") * 100
display(matriz_marcador.round(2).astype(str) + " %")

# 7. Analizamos el impacto del Minuto (Cansancio / Prisas)
print("\n📊 ¿CÓMO CAMBIA EL JUEGO DE NEYMAR SEGÚN EL MINUTO DEL PARTIDO?")
print("-" * 75)
matriz_minuto = pd.crosstab(df_ney_carril["tramo_minuto"], df_ney_carril["desenlace"], normalize="index") * 100
display(matriz_minuto.round(2).astype(str) + " %")
```
